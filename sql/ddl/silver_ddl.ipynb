```python
%python
import os

config_path = os.path.abspath(
    os.path.join(os.getcwd(), "../../config/config.yml")
)

config = load_config(config_path)
```

```python
%python
catalog = config["target"]["catalog"]
silver_schema = config["target"]["silver_schema"]
```

```python
spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{silver_schema}")
```

```python
%python
#============================================================
#ddl_silver_pokemon
#============================================================
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.{silver_schema}.pokemon (
    pokemon_id INT NOT NULL,
    name STRING NOT NULL,
    base_experience INT,
    height INT,
    weight INT,
    is_default BOOLEAN,
    pokemon_order INT,
    species_name STRING,
    species_url STRING,
    source_ingestion_timestamp TIMESTAMP,
    silver_processed_at TIMESTAMP
)
USING DELTA
COMMENT 'Current-state Silver entity table with one row per Pokemon'
""")
```

```python
# ============================================================
# ddl_silver_pokemon_types
# ============================================================
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.{silver_schema}.pokemon_types (
    pokemon_id INT NOT NULL,
    slot INT NOT NULL,
    type_name STRING NOT NULL,
    type_url STRING,
    source_ingestion_timestamp TIMESTAMP,
    silver_processed_at TIMESTAMP
)
USING DELTA
COMMENT 'Silver relationship table with one row per Pokemon type assignment'         
""")
```

```python
# ============================================================
# ddl_silver_pokemon_abilities
# ============================================================
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.{silver_schema}.pokemon_abilities (
    pokemon_id INT NOT NULL,
    slot INT NOT NULL,
    ability_name STRING NOT NULL,
    ability_url STRING,
    is_hidden BOOLEAN,
    source_ingestion_timestamp TIMESTAMP,
    silver_processed_at TIMESTAMP
)
USING DELTA
COMMENT 'Silver relationship table with one row per Pokemon ability assignment'
""")
```

```python
# ============================================================
# ddl_silver_pokemon_stats
# ============================================================
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.{silver_schema}.pokemon_stats (
    pokemon_id INT NOT NULL,
    stat_name STRING NOT NULL,
    stat_url STRING,
    base_stat INT,
    effort INT,
    source_ingestion_timestamp TIMESTAMP,
    silver_processed_at TIMESTAMP
)
USING DELTA
COMMENT 'Silver relationship table with one row per Pokemon stat'
""")
```

```python
# ============================================================
# ddl_silver_pokemon_moves
# ============================================================
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.{silver_schema}.pokemon_moves (
    pokemon_id INT NOT NULL,
    move_name STRING NOT NULL,
    move_url STRING,
    source_ingestion_timestamp TIMESTAMP,
    silver_processed_at TIMESTAMP
)
USING DELTA
COMMENT 'Silver relationship table with one row per Pokemon and move'
""")
```

```python
# ============================================================
# ddl_silver_pokemon_move_details
# ============================================================
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.{silver_schema}.pokemon_move_details (
    pokemon_id INT NOT NULL,
    move_name STRING NOT NULL,
    version_group_name STRING NOT NULL,
    version_group_url STRING,
    learn_method_name STRING NOT NULL,
    learn_method_url STRING,
    level_learned_at INT NOT NULL,
    learn_order INT,
    source_ingestion_timestamp TIMESTAMP,
    silver_processed_at TIMESTAMP
)
USING DELTA
COMMENT 'Silver table with one row per Pokemon move learning detail'
""")
```
